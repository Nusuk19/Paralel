```python
import random
import time
import json

def generate_points(num_points, filename):
    points = [
        {"x": random.uniform(-100, 100), "y": random.uniform(-100, 100), "w": random.uniform(1, 10)}
        for _ in range(num_points)
    ]
    with open(filename, "w") as f:
        json.dump(points, f)

def read_points(filename):
    with open(filename, "r") as f:
        return json.load(f)

def compute_center_of_mass(points):
    sum_wx = sum(p["x"] * p["w"] for p in points)
    sum_wy = sum(p["y"] * p["w"] for p in points)
    sum_w = sum(p["w"] for p in points)
    
    x_c = sum_wx / sum_w
    y_c = sum_wy / sum_w
    
    return x_c, y_c

def main():
    filename = "points.json"
    num_points = 1000000  
    
    generate_points(num_points, filename)
    points = read_points(filename)
    
    start_time = time.time()
    center = compute_center_of_mass(points)
    end_time = time.time()
    
    print(f"Center of Mass: {center}")
    print(f"Execution Time: {end_time - start_time:.5f} seconds")

if __name__ == "__main__":
    main()
```
